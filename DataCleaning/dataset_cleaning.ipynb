# Cleaning "Quotes" and "Sales" Datasets
In this notebook, we loaded, inspected, and cleaned two datasets:
- `Intuilize_MNSU_ACME_QuotesData.csv`
- `Intuilize_MNSU_ACME_SalesData.csv`

The cleaning steps included:
- Loading the data
- Checking for missing values
- Handling duplicates
- Fixing data types





```python
# Importing necessary libraries
import pandas as pd
import numpy as np
```

## Loading the Datasets
We loaded both CSV files into Pandas DataFrames to begin the cleaning process.


```python
# Load the datasets
quotes_df = pd.read_csv('../Dataset/OriginalData/Intuilize_MNSU_ACME_QuotesData.csv')
sales_df = pd.read_csv('../Dataset/OriginalData/Intuilize_MNSU_ACME_SalesData.csv')

# Quick look at the first few rows
quotes_df.head(), sales_df.head()
```

## Checking Dataset Info
We checked the basic information (number of rows, columns, data types, etc.) to understand the structure of each dataset.


```python
# Dataset info
print("Quotes Data Info:")
quotes_df.info()

print("\nSales Data Info:")
sales_df.info()
```

## Checking Missing Values
We checked for any missing (NaN) values in both datasets.


```python
# Check for missing values
print("Missing values in Quotes Data:")
print(quotes_df.isnull().sum())

print("\nMissing values in Sales Data:")
print(sales_df.isnull().sum())
```

## Handling Missing Values
We handled missing data by:
- Dropping columns with too many missing values
- Filling or dropping rows as needed


```python
# Drop columns where more than 50% of the values are missing
quotes_df = quotes_df.dropna(thresh=len(quotes_df) * 0.5, axis=1)
sales_df = sales_df.dropna(thresh=len(sales_df) * 0.5, axis=1)

# Fill remaining missing values
quotes_df = quotes_df.fillna('Unknown')
sales_df = sales_df.fillna('Unknown')
```

## Checking for Duplicates
We checked for duplicated rows in each dataset and removed them if necessary.


```python
# Find and remove duplicates
quotes_duplicates = quotes_df.duplicated().sum()
sales_duplicates = sales_df.duplicated().sum()

print(f"Duplicates in Quotes Data: {quotes_duplicates}")
print(f"Duplicates in Sales Data: {sales_duplicates}")

# Drop duplicates if any
quotes_df = quotes_df.drop_duplicates()
sales_df = sales_df.drop_duplicates()
```

## Fixing Data Types
We corrected any columns that were stored in the wrong format, such as dates or IDs.


```python
# Try to automatically detect and fix dates
for col in quotes_df.columns:
    if 'date' in col.lower():
        quotes_df[col] = pd.to_datetime(quotes_df[col], errors='coerce')

for col in sales_df.columns:
    if 'date' in col.lower():
        sales_df[col] = pd.to_datetime(sales_df[col], errors='coerce')

# Convert ID columns to strings
id_cols = [col for col in quotes_df.columns if 'id' in col.lower()]
for col in id_cols:
    quotes_df[col] = quotes_df[col].astype(str)

id_cols = [col for col in sales_df.columns if 'id' in col.lower()]
for col in id_cols:
    sales_df[col] = sales_df[col].astype(str)
```

## Final Check
We reviewed the datasets one last time to ensure they were clean and ready for analysis.


```python
# Quick summary stats
print("Quotes Data:")
print(quotes_df.describe(include='all'))

print("\nSales Data:")
print(sales_df.describe(include='all'))

```
