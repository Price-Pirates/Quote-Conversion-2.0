# Backward Elimation Method

```python
import pandas as pd
%pip install mlxtend
from mlxtend.feature_selection import SequentialFeatureSelector
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
```

```python
df = pd.read_csv("../Dataset/quotes_customer_tier.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
# Drop the first column
df.drop(columns=df.columns[0], inplace=True)

# Define X and y
X = df.iloc[:, list(range(0, 16)) + list(range(17, 20))]  # Using columns 2-17, 19-21
y = df.iloc[:, 16]  # Column 18 becomes 17 after dropping the first column
```

```python
X.shape
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=1)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
categorical_cols = X.select_dtypes(include=['object']).columns
```

```python
numeric_features = X.select_dtypes(include=['int64', 'float64']).columns
categorical_features = categorical_cols
```

```python
bakward_feature_selector = SequentialFeatureSelector(RandomForestClassifier(n_jobs=-1),
                                                    k_features=(1,20),
                                                    forward=False,
                                                    floating=False,
                                                    verbose=2,
                                                    scoring= "accuracy",
                                                    cv= 5).fit(X_train, y_train)
```

```python
bakward_feature_selector.k_feature_idx_
```
