```python
import pandas as pd

def merge_csv_files(first_file_path, second_file_path, output_file_path, columns_from_second, merge_on_first, merge_on_second):
    # Load the first dataset
    first_df = pd.read_csv(first_file_path)
    
    # Load the second dataset
    second_df = pd.read_csv(second_file_path)
    
    # Ensure the merge column is included in the second dataframe's columns to merge
    required_columns = columns_from_second + [merge_on_second] if merge_on_second not in columns_from_second else columns_from_second
    second_df = second_df[required_columns]

    # Perform a left merge to include all rows from the first dataset and matching rows from the second dataset
    merged_df = pd.merge(first_df, second_df, left_on=merge_on_first, right_on=merge_on_second, how='left')

    # Save the merged dataframe to a new CSV file
    merged_df.to_csv(output_file_path, index=False)

# Example usage
first_file_path = r"C:\Users\shrey\OneDrive\Desktop\original data project\Intuilize_MNSU_ACME_QuotesData.csv"
second_file_path = r"C:\Users\shrey\OneDrive\Desktop\prof.csv"
output_file_path = r"C:\Users\shrey\OneDrive\Desktop\original data project\merged4.csv"
columns_from_second = ['PercentProfit', 'InvoiceNumber', 'LineNumber', 'ExtPrice', 'ExtCost', 'UnitCost']
merge_on_first = 'ConvertedToOrderNumber'
merge_on_second = 'OrderNumber'

merge_csv_files(first_file_path, second_file_path, output_file_path, columns_from_second, merge_on_first, merge_on_second)


```
