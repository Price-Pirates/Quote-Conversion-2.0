```python
import pandas as pd

# Load sales dataset
sales_dataset = pd.read_csv(r"C:\Users\shrey\OneDrive\Desktop\Intuilize_MNSU_ACME_SalesData.csv")

# Keep only EA for UOM
sales_dataset = sales_dataset[sales_dataset['UOM'] == 'EA']

# Calculate Percent Profit
sales_dataset['PercentProfit'] = (((sales_dataset['UnitPrice'] - sales_dataset['UnitCost']) / sales_dataset['UnitCost'])*100)

# Drop duplicates
sales_dataset = sales_dataset.drop_duplicates()

# Save the cleaned sales dataset with profit calculation
sales_dataset.to_csv(r"C:\Users\shrey\OneDrive\Desktop\prof.csv", index=False)

# Display first few rows
print(sales_dataset.head())
```
