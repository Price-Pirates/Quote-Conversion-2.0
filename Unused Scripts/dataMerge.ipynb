```python
""" This is the code that I ran, change the filepath and use it to see how that works out for the full data file. 
My best guess is this is what we are looking for in the processing data?
"""

import pandas as pd

def merge_csv_files(first_file_path, second_file_path, output_file_path, columns_from_second, merge_on_first, merge_on_second):
    # Load the first dataset
    first_df = pd.read_csv(first_file_path)
    
    # Load the second dataset
    second_df = pd.read_csv(second_file_path)
    
    # Convert merge columns to string and strip leading/trailing spaces
    # i am guessing this is what is missing on your current merge file because of empty spaces and irregular formatted data. 
    first_df[merge_on_first] = first_df[merge_on_first].astype(str).str.strip()
    second_df[merge_on_second] = second_df[merge_on_second].astype(str).str.strip()
    
    # Ensure the merge column is included in the second dataframe's columns
    required_columns = columns_from_second + [merge_on_second] if merge_on_second not in columns_from_second else columns_from_second
    second_df = second_df[required_columns]

    # Perform a left merge to include all rows from the first dataset and matching rows from the second dataset
    merged_df = pd.merge(first_df, second_df, left_on=merge_on_first, right_on=merge_on_second, how='left')

    # Save the merged dataframe to a new CSV file
    merged_df.to_csv(output_file_path, index=False)
    print(f'Merged data saved to: {output_file_path}')

# Example usage
first_file_path = r"/Users/roshit/Desktop/Quote-Conversion-2.0-main/Data/OriginalData/Daataa/Intuilize_MNSU_ACME_QuotesData.csv"
second_file_path = r"/Users/roshit/Desktop/Quote-Conversion-2.0-main/Data/OriginalData/Daataa/SalesData_Profit.csv"
output_file_path = r"/Users/roshit/Desktop/Quote-Conversion-2.0-main/Data/OriginalData/Daataa/merged4new.csv"
columns_from_second = ['PercentProfit', 'InvoiceNumber', 'LineNumber', 'ExtPrice', 'ExtCost', 'UnitCost']
merge_on_first = 'ConvertedToOrderNumber'
merge_on_second = 'OrderNumber'

merge_csv_files(first_file_path, second_file_path, output_file_path, columns_from_second, merge_on_first, merge_on_second)
```
